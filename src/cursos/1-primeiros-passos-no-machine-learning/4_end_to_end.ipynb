## Treinar o modelo

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
import pickle
```

```python
# Carregar a base de dados Iris
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'
colunas = ['SepalLength', 'SepalWidth', 'PetalLength', 'PetalWidth', 'Class']
df = pd.read_csv(url, header=None, names=colunas)
```

```python
# Separar as variáveis independentes (X) e dependentes (y)
X = df[['SepalLength', 'SepalWidth', 'PetalLength', 'PetalWidth']]  # Features
y = df['Class']
```

```python
# Dividir os dados em treino e teste
semente = 42
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=semente)
```

```python
# Criar e treinar o modelo de regressão logística com 100 árvores
modelo = RandomForestClassifier(n_estimators=100, random_state=semente)
modelo.fit(X_train, y_train)
```

```python
# Avaliar o modelo
score = modelo.score(X_test, y_test)
print(f'Acurácia do modelo: {score:.4f}')
```

## Salvar o binário (objeto serializado) do modelo

```python
# Salvar o modelo em um arquivo pickle
with open('modelo_classificacao.pkl', 'wb') as f:
    pickle.dump(modelo, f)
```

## Carregar o binário

```python
with open('/content/modelo_classificacao.pkl', 'rb') as f:
    modelo_carregado = pickle.load(f)
```
