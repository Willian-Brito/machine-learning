# Quarteto de Anscombe

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import linregress
```

```python
# Dados do Quarteto de Anscombe
anscombe_data = {
    "dataset": ["I"] * 11 + ["II"] * 11 + ["III"] * 11 + ["IV"] * 11,
    "x": [
        10, 8, 13, 9, 11, 14, 6, 4, 12, 7, 5,  # Dataset I
        10, 8, 13, 9, 11, 14, 6, 4, 12, 7, 5,  # Dataset II
        10, 8, 13, 9, 11, 14, 6, 4, 12, 7, 5,  # Dataset III
        8, 8, 8, 8, 8, 8, 8, 19, 8, 8, 8        # Dataset IV
    ],
    "y": [
        8.04, 6.95, 7.58, 8.81, 8.33, 9.96, 7.24, 4.26, 10.84, 4.82, 5.68,  # Dataset I
        9.14, 8.14, 8.74, 8.77, 9.26, 8.10, 6.13, 3.10, 9.13, 7.26, 4.74,   # Dataset II
        7.46, 6.77, 12.74, 7.11, 7.81, 8.84, 6.08, 5.39, 8.15, 6.42, 5.73,  # Dataset III
        6.58, 5.76, 7.71, 8.84, 8.47, 7.04, 5.25, 12.50, 5.56, 7.91, 6.89   # Dataset IV
    ]
}

# Criando um DataFrame
anscombe_df = pd.DataFrame(anscombe_data)
```

```python
# Função para calcular estatísticas resumidas
def calcular_estatisticas(df):
    stats = []
    for dataset in df["dataset"].unique():
        subset = df[df["dataset"] == dataset]
        x = subset["x"]
        y = subset["y"]
        slope, intercept, r_value, _, _ = linregress(x, y)
        stats.append({
            "Dataset": dataset,
            "Média X": np.mean(x),
            "Média Y": np.mean(y),
            "Variancia X": np.var(x, ddof=1),
            "Variancia Y": np.var(y, ddof=1),
            "Correlação": r_value,
            "Inclinação": slope,
            "Interceptação": intercept
        })
    return pd.DataFrame(stats)
```

```python
# Função para plotar os gráficos
def plot_anscombe(df):
    fig, axes = plt.subplots(2, 2, figsize=(10, 8), sharex=True, sharey=True)
    datasets = df["dataset"].unique()
    for i, dataset in enumerate(datasets):
        subset = df[df["dataset"] == dataset]
        ax = axes[i // 2, i % 2]
        ax.scatter(subset["x"], subset["y"], label=f"Dataset {dataset}")
        slope, intercept, _, _, _ = linregress(subset["x"], subset["y"])
        x_vals = np.linspace(min(subset["x"]), max(subset["x"]), 100)
        y_vals = slope * x_vals + intercept
        ax.plot(x_vals, y_vals, color="red", label="Regressão Linear")
        ax.set_title(f"Dataset {dataset}")
        ax.legend()
    fig.suptitle("Quarteto de Anscombe", fontsize=16)
    plt.tight_layout()
    plt.show()
```

```python
# Calculando estatísticas
statistics = calcular_estatisticas(anscombe_df)
print("Estatísticas Resumidas:\n", statistics)
```

```python
# Plotando os gráficos
plot_anscombe(anscombe_df)
```

# Medidas de tendência central

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Criando um conjunto de dados fictício
data = {
    "Grupo": ["A"] * 10 + ["B"] * 10 + ["C"] * 10,
    "Valores": [
        10, 12, 14, 15, 10, 10, 11, 13, 15, 10,  # Grupo A
        20, 22, 19, 18, 21, 22, 20, 19, 21, 20,  # Grupo B
        30, 32, 31, 33, 30, 30, 29, 35, 34, 31   # Grupo C
    ]
}

# Criando um DataFrame
df = pd.DataFrame(data)
```

```python
# Função para calcular medidas de tendência central
def calculo_tendencia_central(df):
    stats = []
    for group in df["Grupo"].unique():
        subset = df[df["Grupo"] == group]
        valores = subset["Valores"]
        stats.append({
            "Grupo": group,
            "Média": np.mean(valores),
            "Mediana": np.median(valores),
            "Moda": valores.mode().iloc[0]
        })
    return pd.DataFrame(stats)
```

```python
# Visualizando os dados e as medidas de tendência central
def plot_tendencia_central(df):
    grupos = df["Grupo"].unique()
    fig, axes = plt.subplots(1, len(grupos), figsize=(15, 5), sharey=True)
    for i, group in enumerate(grupos):
        subset = df[df["Grupo"] == group]
        valores = subset["Valores"]
        ax = axes[i]
        ax.hist(valores, bins=5, color="skyblue", edgecolor="black", alpha=0.7)
        ax.axvline(np.mean(valores), color="red", linestyle="-", label="Média")
        ax.axvline(np.median(valores), color="green", linestyle="--", label="Mediana")
        moda = valores.mode().iloc[0]
        ax.axvline(moda, color="blue", linestyle="-.", label="Moda")
        ax.set_title(f"Grupo {group}")
        ax.set_xlabel("Valores")
        ax.set_ylabel("Frequência")
        ax.legend()
    plt.suptitle("Medidas de Tendência Central por Grupo", fontsize=16)
    plt.tight_layout()
    plt.show()
```

```python
# Calculando as medidas de tendência central
print("Medidas de Tendência Central:\n", calculo_tendencia_central(df))
```

```python
# Plotando os gráficos
plot_tendencia_central(df)
```

# Medidas de dispersão

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Criando um conjunto de dados fictício
data = {
    "Grupo": ["A"] * 10 + ["B"] * 10 + ["C"] * 10,
    "Valores": [
        10, 12, 14, 15, 10, 10, 11, 13, 15, 10,  # Grupo A
        20, 22, 19, 18, 21, 22, 20, 19, 21, 20,  # Grupo B
        30, 32, 31, 33, 30, 30, 29, 35, 34, 31   # Grupo C
    ]
}

# Criando um DataFrame
df = pd.DataFrame(data)
```

```python
# Função para calcular medidas de dispersão
def calculo_medida_dispersao(df):
    stats = []
    for group in df["Grupo"].unique():
        subset = df[df["Grupo"] == group]
        valores = subset["Valores"]
        stats.append({
            "Grupo": group,
            "Variância": np.var(valores, ddof=1),
            "Desvio Padrão": np.std(valores, ddof=1),
            "Amplitude": np.max(valores) - np.min(valores),
            "Coeficiente de Variação (%)": (np.std(valores, ddof=1) / np.mean(valores)) * 100
        })
    return pd.DataFrame(stats)
```

```python
# Calculando as medidas de dispersão
print("Medidas de Dispersão:\n", calculo_medida_dispersao(df))
```

```python
# Medidas de tendência central e dispersão por coluna numérica
def estatistica_detalhada(df):
    stats = {}
    for col in df.select_dtypes(include=[np.number]):
        stats[col] = {
            "Média": np.mean(df[col]),
            "Mediana": np.median(df[col]),
            "Desvio Padrão": np.std(df[col], ddof=1),
            "Amplitude": np.max(df[col]) - np.min(df[col])
        }
    print("\nMedidas de Tendência Central e Dispersão por coluna:")
    for col, measures in stats.items():
        print(f"\n{col}:")
        for measure, value in measures.items():
            print(f"  {measure}: {value:.2f}")
```

# Probabilidade Condicional

```python
import random
```

```python
# Simulação de dois dados de 6 faces
def lancar_dados():
    return random.randint(1, 6), random.randint(1, 6)
```

```python
# Espaço amostral dos dois dados
espaco_amostral = [(d1, d2) for d1 in range(1, 7) for d2 in range(1, 7)]
```

```python
# Eventos de interesse
# Primeiro dado ímpar
evento_primeiro_impar = [(d1, d2) for d1, d2 in espaco_amostral if d1 % 2 != 0]

# Ambos: primeiro dado ímpar e segundo dado ímpar
evento_primeiro_e_segundo_impar = [(d1, d2) for d1, d2 in evento_primeiro_impar if d2 % 2 != 0]

# Probabilidade condicional: P(Segundo dado ímpar | Primeiro dado ímpar)
probabilidade_condicional = len(evento_primeiro_e_segundo_impar) / len(evento_primeiro_impar)
```

```python
print("Espaço amostral (todos os lançamentos possíveis):", len(espaco_amostral))
print("Eventos com o primeiro dado ímpar:", len(evento_primeiro_impar))
print("Eventos com ambos ímpares:", len(evento_primeiro_e_segundo_impar))
print(f"\nProbabilidade condicional (P(Segundo dado ímpar | Primeiro dado ímpar)): {probabilidade_condicional * 100:.2f}%")
```

```python
# Simulação de lançamentos para validação
n_lancamentos = 10000
lancamentos = [lancar_dados() for _ in range(n_lancamentos)]
```

```python
# Filtros para eventos da simulação
sim_primeiro_impar = [l for l in lancamentos if l[0] % 2 != 0]
sim_ambos_impares = [l for l in sim_primeiro_impar if l[1] % 2 != 0]

# Probabilidade condicional na simulação
sim_probabilidade_condicional = len(sim_ambos_impares) / len(sim_primeiro_impar)
```

```python
print(f"\nSimulação com {n_lancamentos} lançamentos:")
print(f"Probabilidade condicional observada: {sim_probabilidade_condicional * 100:.2f}%")
```

# Teorema de Bayes

```python
# Definir as probabilidades conhecidas
P_Doenca = 0.01  # Probabilidade a priori de ter a doença
P_SemDoenca = 1 - P_Doenca  # Complemento (não ter a doença)
```

```python
P_TestePos_Doenca = 0.95  # Probabilidade de um teste positivo dado que a pessoa tem a doença
P_TestePos_SemDoenca = 0.10  # Probabilidade de um teste positivo dado que a pessoa não tem a doença
```

```python
# Calcular P(TestePositivo)
P_TestePos = (P_TestePos_Doenca * P_Doenca) + (P_TestePos_SemDoenca * P_SemDoenca)
```

```python
# Aplicar o Teorema de Bayes para calcular P(Doença | TestePositivo)
P_Doenca_TestePos = (P_TestePos_Doenca * P_Doenca) / P_TestePos
```

```python
# Exibir o resultado
print(f"A probabilidade de ter a doença, dado que o teste foi positivo é: {P_Doenca_TestePos*100:.2f}%")
```

# Teste de Hipóteses

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
```

```python
# Gerando dois conjuntos de dados fictícios
np.random.seed(42)
grupo_a = np.random.normal(100, 15, 30)  # Média 100, desvio padrão 15, tamanho 30
grupo_b = np.random.normal(105, 15, 30)  # Média 105, desvio padrão 15, tamanho 30
#grupo_b = np.random.normal(305, 15, 30)  # Média 105, desvio padrão 15, tamanho 30

```

Hipótese:

**H0 (hipótese nula):** A média do Grupo A é igual à média do Grupo B (não há diferença significativa).

**Ha (hipótese alternativa):** A média do Grupo A é diferente da média do Grupo B (há diferença significativa).

## T-Test

```python
# Realizando o T-Test
t_stat, p_value = stats.ttest_ind(grupo_a, grupo_b)

# Resultados do T-Test
print("Resultados do T-Test:")
print(f"Estatística t: {t_stat:.4f}")
print(f"P-Value: {p_value:.4f}")
```

## P-Value

```python
# Interpretação do P-Value
alfa = 0.05  # Nível de significância
if p_value < alfa:
    print(f"\nComo o P-Value ({p_value:.4f}) é menor que {alfa}, rejeitamos a hipótese nula (H0).\n")
else:
    print(f"\nComo o P-Value ({p_value:.4f}) é maior ou igual a {alfa}, *NÃO* rejeitamos a hipótese nula (H0).\n")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
# Gerando dados de exemplo
np.random.seed(42)
x = np.random.rand(100, 1) * 10  # Variável independente (100 valores entre 0 e 10)
y = 2.5 * x + np.random.randn(100, 1) * 2  # Variável dependente com ruído
```

```python
# Realizando o Cross-Validation manualmente
for i, (train_index, test_index) in enumerate(kf.split(X)):
    X_train, X_test = X[train_index], X[test_index]
    y_train, y_test = y[train_index], y[test_index]

    # Treinando o modelo com os dados de treino
    modelo.fit(X_train, y_train)

    # Avaliando o modelo nos dados de teste
    score = modelo.score(X_test, y_test)  # R²
    scores.append(score)

    print(f"Pasta {i + 1}: R^2 = {score:.4f}")
```

```python
# Calculando a média e o desvio padrão dos scores
mean_score = np.mean(scores)
std_score = np.std(scores)

print(f"\nMédia do R^2 após {k} folds: {mean_score:.4f}")
print(f"Desvio padrão do R^2: {std_score:.4f}")
```

```python
# Alternativa: Usando cross_val_score diretamente
cv_scores = cross_val_score(modelo, X, y, cv=kf, scoring='r2')

print("\nResultados com cross_val_score:")
print(f"R^2 por pasta: {cv_scores}")
print(f"Média do R^2: {cv_scores.mean():.4f}")
print(f"Desvio padrão do R^2: {cv_scores.std():.4f}")
```
